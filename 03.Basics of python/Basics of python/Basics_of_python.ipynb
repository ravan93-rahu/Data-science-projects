**Exercise 1: Prime Numbers**

Write a Python program that checks whether a given number is prime or not. A prime number is a natural number greater than 1 that has no positive divisors other than 1 and itself.

```python
num = int(input("Enter a number: "))

if num <= 1:
    print("Not a prime number")
else:
    for i in range(2, int(num**0.5) + 1):
        if num % i == 0:
            print("Not a prime number")
            break
    else:
        print("Prime number")
```

**Exercise 2: Product of Random Numbers**

Develop a Python program that generates two random numbers and asks the user to enter the product of these numbers. The program should then check if the user's answer is correct and display an appropriate message.






1.  **Generating Random Numbers**

2.  **Calculating the Product**
3.  **Getting User Input and Checking the Answer**


```python
import random

num1 = random.randint(1, 10) # Generate a random integer between 1 and 10
num2 = random.randint(1, 10) # Generate another random integer between 1 and 10

print(f"The numbers are: {num1} and {num2}")
```

```python
product = num1 * num2
print(f"The product of {num1} and {num2} is {product}")
```

```python
user = int(input("Enter the product of the two numbers: "))

if user == product:
    print("Correct! Your answer is right.")
else:
    print(f"Incorrect. The correct answer is {product}.")
```

**Exercise 4: Word counter**

write a program to count the number of words in a given text.
example:
input_text = "This is a sample text. This text will be used to demonstrate the word counter."
Expected output:
'This': 2
'is': 1
'a': 1
'sample': 1
'text.': 1


```python
#1. take input(text) from user
user_text=input("This is a sample text. This text will be used to demonstrate the word counter")

#2.remove punctuation
without_dot=user_text.replace('.' , '')
end_clean_text=without_dot.replace(',','')

#3.crate list of text
words_list=end_clean_text.lower().split()


#4.create dict
frequency_dict={}

for word in words_list:
  if word in frequency_dict:
    frequency_dict[word]+=1

  else:
    frequency_dict[word]=1

print(frequency_dict)

```

**Exercise 5: Check for Palindrome**

Write a Python function called is_palindrome that takes a string as input and returns True if the string is a palindrome, and False otherwise. A palindrome is a word, phrase, number, or other sequence of characters that reads the same forward and backward, ignoring spaces, punctuation, and capitalization.

```python
import re    ##(regular expression)

def is_palindrome(text):
  # Remove non-alphanumeric characters and convert to lowercase
  cleaned_text = re.sub(r'[^a-zA-Z0-9]', '', text).lower()
  reversed_text = cleaned_text[::-1]

  if cleaned_text == reversed_text:
    return True
  else:
    return False

# Example usage:
print(is_palindrome("Racecar"))
```
