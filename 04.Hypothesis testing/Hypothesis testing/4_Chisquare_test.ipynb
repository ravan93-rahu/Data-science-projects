### 1. State the Hypotheses

*   **Null Hypothesis (H0):** There is no significant association between the type of smart home device purchased (Smart Thermostats vs. Smart Lights) and the customer satisfaction level. The variables are independent.
*   **Alternative Hypothesis (H1):** There is a significant association between the type of smart home device purchased (Smart Thermostats vs. Smart Lights) and the customer satisfaction level. The variables are dependent.

```python
import pandas as pd
import numpy as np
from scipy.stats import chi2_contingency, chi2

# Data Provided: Contingency Table
data = {
    'Smart Thermostat': [50, 80, 60, 30, 20],
    'Smart Light': [70, 100, 90, 50, 50]
}
index = ['Very Satisfied', 'Satisfied', 'Neutral', 'Unsatisfied', 'Very Unsatisfied']
contingency_table = pd.DataFrame(data, index=index)

print("Contingency Table:")
display(contingency_table)
```

### 2. Compute the Chi-Square Statistic

We will use `scipy.stats.chi2_contingency` to compute the Chi-Square statistic, p-value, degrees of freedom, and expected frequencies.

```python
# Perform the Chi-Square test for independence
chi2_stat, p_value, dof, expected_frequencies = chi2_contingency(contingency_table)

print(f"Chi-Square Statistic: {chi2_stat:.4f}")
print(f"P-value: {p_value:.4f}")
print(f"Degrees of Freedom (dof): {dof}")

print("\nExpected Frequencies Table:")
expected_df = pd.DataFrame(expected_frequencies, index=contingency_table.index, columns=contingency_table.columns)
display(expected_df)
```

### 3. Determine the Critical Value

Using a significance level (alpha) of 0.05 and the calculated degrees of freedom, we will find the critical Chi-Square value.

```python
alpha = 0.05
critical_value = chi2.ppf(1 - alpha, dof)

print(f"Significance Level (alpha): {alpha}")
print(f"Degrees of Freedom (dof): {dof}")
print(f"Critical Value: {critical_value:.4f}")
```

### 4. Make a Decision

We compare the calculated Chi-Square statistic with the critical value and the p-value with the significance level to decide whether to reject the null hypothesis.

```python
print(f"Chi-Square Statistic: {chi2_stat:.4f}")
print(f"Critical Value: {critical_value:.4f}")
print(f"P-value: {p_value:.4f}")
print(f"Significance Level (alpha): {alpha}")

if chi2_stat > critical_value:
    print("\nDecision (Chi-Square Statistic vs Critical Value): Reject the Null Hypothesis.")
else:
    print("\nDecision (Chi-Square Statistic vs Critical Value): Fail to Reject the Null Hypothesis.")

if p_value < alpha:
    print("Decision (P-value vs Alpha): Reject the Null Hypothesis.")
else:
    print("Decision (P-value vs Alpha): Fail to Reject the Null Hypothesis.")

print("\nConclusion: Based on the Chi-Square test, since the Chi-Square statistic is greater than the critical value and the p-value is less than the significance level (alpha = 0.05), we reject the null hypothesis. This means there is a statistically significant association between the type of smart home device purchased and the customer satisfaction level.")
```
