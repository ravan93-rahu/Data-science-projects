```python
import pandas as pd

# Load the dataset
df = pd.read_csv('/content/anime.csv')

# Display the first 5 rows of the DataFrame
display(df.head())
```

```python
# Check for missing values
print("Missing values in each column:")
display(df.isnull().sum())

# Display data types and non-null values
print("\nDataFrame Info:")
display(df.info())
```

```python
# Fill missing 'genre' and 'type' values with 'Unknown'
df['genre'].fillna('Unknown', inplace=True)
df['type'].fillna('Unknown', inplace=True)

# Fill missing 'rating' values with the mean of the column
df['rating'].fillna(df['rating'].mean(), inplace=True)

# Inspect unique values in 'episodes' column to identify non-numeric entries
print("Unique values in 'episodes' column before conversion:")
display(df['episodes'].unique())

# Replace 'Unknown' with NaN in 'episodes' to allow conversion to numeric, then fill with 1
df['episodes'] = df['episodes'].replace('Unknown', '1') # Assuming 'Unknown' episodes can be treated as 1 episode or handle as appropriate

# Convert 'episodes' to numeric (float), coercing errors to NaN
df['episodes'] = pd.to_numeric(df['episodes'], errors='coerce')

# Fill any NaN values introduced by coercion (e.g., if there were other non-numeric strings) with 1
df['episodes'].fillna(1, inplace=True)

# Convert 'episodes' to integer type after handling NaNs
df['episodes'] = df['episodes'].astype(int)


print("\nMissing values after handling:")
display(df.isnull().sum())

print("\nDataFrame Info after preprocessing:")
display(df.info())
```

```python
# Clean and split genres
df['genre'] = df['genre'].apply(lambda x: [s.strip() for s in x.split(',')])

# Display the first few entries of the processed genre column
print("Processed 'genre' column for the first 5 rows:")
display(df['genre'].head())
```

```python
from sklearn.feature_extraction.text import CountVectorizer

# Create a CountVectorizer to convert genres into a matrix
# Join the list of genres back into a single string for CountVectorizer
df['genre_string'] = df['genre'].apply(lambda x: ' '.join(x))

cv = CountVectorizer()
genre_matrix = cv.fit_transform(df['genre_string'])

# Display the shape of the genre matrix
print("Shape of the genre matrix:", genre_matrix.shape)

# Display the first few feature names (genres)
print("\nFirst 10 genre features:", cv.get_feature_names_out()[:10])
```

```python
from sklearn.metrics.pairwise import cosine_similarity

# Compute cosine similarity matrix
cosine_sim = cosine_similarity(genre_matrix)

# Display the shape of the cosine similarity matrix
print("Shape of the cosine similarity matrix:", cosine_sim.shape)

# Display a small part of the similarity matrix (e.g., first 5x5)
print("\nFirst 5x5 of the cosine similarity matrix:")
display(pd.DataFrame(cosine_sim[:5, :5]))
```

```python
# Drop duplicate anime names to avoid issues with indexing
anime_df = df.drop_duplicates(subset=['name'])

# Reset index after dropping duplicates
anime_df = anime_df.reset_index(drop=True)

# Create a Series for mapping anime titles to their indices
titles = pd.Series(anime_df.index, index=anime_df['name']).drop_duplicates()

# Define the recommendation function
def get_recommendations(title, cosine_sim_matrix, df_anime, top_n=10):
    # Get the index of the anime that matches the title
    if title not in titles:
        print(f"Anime '{title}' not found in the dataset.")
        return []

    idx = titles[title]

    # Get the pairwise similarity scores of all anime with that anime
    sim_scores = list(enumerate(cosine_sim_matrix[idx]))

    # Sort the anime based on the similarity scores
    sim_scores = sorted(sim_scores, key=lambda x: x[1], reverse=True)

    # Get the scores of the top_n most similar anime
    sim_scores = sim_scores[1:top_n+1]  # Exclude itself

    # Get the anime indices
    anime_indices = [i[0] for i in sim_scores]

    # Return the top_n most similar anime names
    return df_anime['name'].iloc[anime_indices].tolist()

# Test the recommendation system
print("Recommendations for 'Kimi no Na wa.':")
display(get_recommendations('Kimi no Na wa.', cosine_sim, anime_df))

print("\nRecommendations for 'Naruto':")
display(get_recommendations('Naruto', cosine_sim, anime_df))
```

## Interview Questions and Answers

### 1. Can you explain the difference between user-based and item-based collaborative filtering?

**User-Based Collaborative Filtering:**
*   **Concept:** Recommends items to a user based on the preferences of other users who are *similar* to them.
*   **How it works:** It finds a group of users (neighbors) with similar tastes to the target user. Then, it recommends items that these similar users have liked but the target user has not yet experienced.
*   **Strengths:** Can recommend diverse items and capture serendipity.
*   **Weaknesses:** Suffers from the 'cold start' problem for new users (hard to find similar users), scalability issues with many users, and sparsity of user-item interaction data.

**Item-Based Collaborative Filtering:**
*   **Concept:** Recommends items to a user based on the items that are *similar* to the ones they have already liked.
*   **How it works:** It identifies items that are similar to each other. When a user interacts with an item, the system recommends other items that are highly similar to the interacted item.
*   **Strengths:** Generally more scalable and robust than user-based filtering for large datasets, as item similarity tends to be more stable over time than user tastes.
*   **Weaknesses:** Can suffer from the 'cold start' problem for new items (hard to find similar items if no one has interacted with them).

### 2. What is collaborative filtering, and how does it work?

**Collaborative Filtering (CF):**
*   **Definition:** Collaborative filtering is a technique used by recommendation systems to make automatic predictions (filtering) about the interests of a user by collecting preferences or taste information from many users (collaborating).
*   **Core Idea:** The fundamental assumption behind CF is that if two users have similar opinions on a set of items, they are likely to have similar opinions on other items as well. Or, if two items are similar, users who like one are likely to like the other.
*   **How it works (General Steps):**
    1.  **Collect Data:** Gather explicit (e.g., ratings) or implicit (e.g., views, purchases) user interaction data for items.
    2.  **Find Similarities:** Calculate similarity between users (user-based CF) or between items (item-based CF). Common similarity metrics include Pearson correlation, cosine similarity, or Jaccard similarity.
    3.  **Generate Recommendations:**
        *   **User-Based:** For a target user, identify a set of most similar users. Then, predict the target user's rating for unrated items by taking a weighted average of the similar users' ratings for those items.
        *   **Item-Based:** For an item a target user has interacted with, identify items most similar to it. Then, predict the target user's rating for an unrated item by taking a weighted average of the ratings the user gave to similar items.
    4.  **Rank and Present:** The items with the highest predicted ratings or similarity scores are then ranked and presented as recommendations to the user.

Collaborative filtering is widely used in many applications, from movie and music recommendations to e-commerce product suggestions.
